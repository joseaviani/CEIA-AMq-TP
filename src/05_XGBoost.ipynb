2. Outliers

Gradient Boosting y Random Forest son bastante robustos a outliers, mucho más que modelos lineales.

Sin embargo, en tu dataset hay colas largas en algunas variables (residual sugar, chlorides, total sulfur dioxide).

Eso puede hacer que el modelo "gaste árboles" tratando de separar instancias muy extremas.

Opciones si querés tratarlos:

Winsorizing: recortar valores extremos a un percentil (ej. 1% y 99%).

Transformaciones logarítmicas: útil en variables muy sesgadas (residual sugar, chlorides).

Nada: si usás boosting, muchas veces se prefiere dejarlos y confiar en que los árboles manejan bien esas colas.

👉 Recomendación práctica:
Hacé un EDA con boxplots/histogramas y probá entrenar el modelo con y sin tratamiento de outliers. No es obligatorio, pero puede mejorar la estabilidad.
